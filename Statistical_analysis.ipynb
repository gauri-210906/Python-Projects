Project: Statistical Analysis of California Housing Prices

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing(as_frame=True)

df = housing.frame

print(df.head())
```

```python
print(df.info())
print(df.describe())
```

```python
print("Mean\n")
print(df.mean())

print("\nMedian\n")
print(df.median())

print("\nStandard Deviation\n")
print(df.std())
```

```python
corr = df.corr()

print(corr)
```

```python
# Plot House Price Distribution
plt.figure(figsize=(7,5))

plt.hist(df["MedHouseVal"], bins=30)

plt.title("Distribution of House Prices")
plt.xlabel("Median House Value")
plt.ylabel("Frequency")

plt.show()
```

```python
# Plot Income Distribution
plt.figure(figsize=(7,5))

plt.hist(df["MedInc"], bins=30)

plt.title("Distribution of Median Income")
plt.xlabel("Median Income")
plt.ylabel("Frequency")

plt.show()
```

```python
# Scatter Plot (Income vs House Price)
plt.figure(figsize=(7,5))

plt.scatter(df["MedInc"], df["MedHouseVal"], alpha=0.4)

plt.xlabel("Median Income")
plt.ylabel("Median House Value")

plt.title("Income vs House Price")

plt.show()
```

```python
# find strong correlation 
print(
    corr["MedHouseVal"]
    .sort_values(ascending=False)
)
```
