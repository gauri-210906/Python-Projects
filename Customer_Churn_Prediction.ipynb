```python
# ============================================================
# CUSTOMER CHURN PREDICTION
# Classification using Logistic Regression
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

```python
# ============================================================
# 2. LOAD DATASET
# ============================================================

# Make sure the CSV file is in the same folder as your notebook

df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

print("Dataset loaded successfully!")
```

```python
# ============================================================
# 3. UNDERSTAND THE DATASET
# ============================================================

print("\n================ FIRST 5 ROWS ================")
print(df.head())

print("\n================ LAST 5 ROWS ================")
print(df.tail())

print("\n================ DATASET SHAPE ================")
print(df.shape)

print("\n================ COLUMN NAMES ================")
print(df.columns.tolist())

print("\n================ DATA TYPES ================")
print(df.dtypes)

print("\n================ DATASET INFO ================")
df.info()

print("\n================ STATISTICAL SUMMARY ================")
print(df.describe())
```

```python
# ============================================================
# 4. CHECK MISSING VALUES
# ============================================================

print("\n================ MISSING VALUES ================")

print(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())
```

```python
# ============================================================
# 5. DATA CLEANING
# ============================================================

df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

print("\nMissing values after converting TotalCharges:")
print(df.isnull().sum())


# Fill missing TotalCharges using median
df["TotalCharges"] = df["TotalCharges"].fillna(
    df["TotalCharges"].median()
)

# Check again
print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nTotal missing values after cleaning:")
print(df.isnull().sum().sum())
```

```python
# ============================================================
# 6. CHECK DUPLICATE RECORDS
# ============================================================

print("\n================ DUPLICATES ================")

print("Number of duplicate rows:")

print(df.duplicated().sum())


# Remove duplicates if present

df = df.drop_duplicates()

print("\nShape after removing duplicates:")

print(df.shape)
```

```python
# ============================================================
# 7. CUSTOMER ID
# ============================================================

# CustomerID is an identifier.
# It does not provide useful information for prediction.

print("\nUnique Customer IDs:")

print(df["CustomerID"].nunique())

# Remove CustomerID before ML

df = df.drop("CustomerID", axis=1)
```

```python
# ============================================================
# 8. TARGET VARIABLE - CHURN
# ============================================================

print("\n================ CHURN VALUES ================")

print(df["Churn"].value_counts())


# Convert:
# Yes -> 1
# No  -> 0

df["Churn"] = df["Churn"].map({
    "Yes": 1,
    "No": 0
})


print("\nChurn after encoding:")

print(df["Churn"].value_counts())
```

```python
# ============================================================
# 9. CHURN PERCENTAGE
# ============================================================

print("\n================ CHURN PERCENTAGE ================")

churn_percentage = df["Churn"].value_counts(
    normalize=True
) * 100

print(churn_percentage)
```

```python
# ============================================================
# 20. CORRELATION ANALYSIS
# ============================================================

numeric_columns = [
    "Age",
    "Tenure",
    "MonthlyCharges",
    "TotalCharges",
    "Churn"
]

correlation = df[numeric_columns].corr()

print("\n================ CORRELATION MATRIX ================")

print(correlation)
```

```python
# ============================================================
# 22. SEPARATE FEATURES AND TARGET
# ============================================================

X = df.drop("Churn", axis=1)

y = df["Churn"]

print("\n================ FEATURES ================")

print(X.head())

print("\nTarget:")

print(y.head())
```

```python
# ============================================================
# 23. ENCODE CATEGORICAL VARIABLES
# ============================================================

print("\nCategorical columns:")

print(
    X.select_dtypes(
        include="object"
    ).columns.tolist()
)


# One-hot encoding

X = pd.get_dummies(
    X,
    drop_first=True,
    dtype=int
)


print("\n================ AFTER ONE-HOT ENCODING ================")

print(X.head())

print("\nNumber of features:")

print(X.shape[1])
```

```python
# ============================================================
# 24. TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


print("\n================ TRAIN TEST SPLIT ================")

print("Training data:", X_train.shape)

print("Testing data :", X_test.shape)

print("Training target:", y_train.shape)

print("Testing target :", y_test.shape)
```

```python
# ============================================================
# 25. FEATURE SCALING
# ============================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train
)

X_test_scaled = scaler.transform(
    X_test
)

print("\nFeature scaling completed.")
```

```python
# ============================================================
# 26. LOGISTIC REGRESSION MODEL
# ============================================================

model = LogisticRegression(
    max_iter=1000
)


# Train model

model.fit(
    X_train_scaled,
    y_train
)


print("\n================ MODEL TRAINING ================")

print("Logistic Regression trained successfully!")
```

```python
# ============================================================
# 27. MAKE PREDICTIONS
# ============================================================

y_pred = model.predict(
    X_test_scaled
)


print("\n================ PREDICTIONS ================")

print("Actual values:")

print(y_test.values[:20])

print("\nPredicted values:")

print(y_pred[:20])
```

```python
# ============================================================
# 28. PREDICT CHURN PROBABILITY
# ============================================================

y_probability = model.predict_proba(
    X_test_scaled
)[:, 1]


print("\n================ CHURN PROBABILITY ================")

print(y_probability[:10])
```

```python
# ============================================================
# 29. ACCURACY
# ============================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)


# ============================================================
# 30. PRECISION
# ============================================================

precision = precision_score(
    y_test,
    y_pred
)

# ============================================================
# 31. RECALL
# ============================================================

recall = recall_score(
    y_test,
    y_pred
)


# ============================================================
# 32. F1 SCORE
# ============================================================

f1 = f1_score(
    y_test,
    y_pred
)

print("\n============================================================")
print("                 MODEL PERFORMANCE")
print("============================================================")

print(f"Accuracy  : {accuracy:.4f}")

print(f"Precision : {precision:.4f}")

print(f"Recall    : {recall:.4f}")

print(f"F1 Score  : {f1:.4f}")
```

```python
# ============================================================
# 34. CLASSIFICATION REPORT
# ============================================================

print("\n================ CLASSIFICATION REPORT ================")

print(
    classification_report(
        y_test,
        y_pred
    )
)
```

```python
# ============================================================
# 35. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred
)
print("\n================ CONFUSION MATRIX ================")
```

```python
# ============================================================
# 37. EXTRACT TN, FP, FN, TP
# ============================================================

tn, fp, fn, tp = cm.ravel()

print("\n================ CONFUSION MATRIX DETAILS ================")
print("True Negative  :", tn)
print("False Positive :", fp)
print("False Negative :", fn)
print("True Positive  :", tp)
```

```python
# ============================================================
# 38. MANUAL METRIC CALCULATIONS
# ============================================================

manual_accuracy = (
    tp + tn
) / (
    tp + tn + fp + fn
)

manual_precision = tp / (
    tp + fp
)

manual_recall = tp / (
    tp + fn
)

manual_f1 = (
    2 * manual_precision * manual_recall
) / (
    manual_precision + manual_recall
)

print("\n================ MANUAL CALCULATIONS ================")
print(
    f"Accuracy  : {manual_accuracy:.4f}"
)
print(
    f"Precision : {manual_precision:.4f}"
)
print(
    f"Recall    : {manual_recall:.4f}"
)

print(
    f"F1 Score  : {manual_f1:.4f}"
)
```

```python
# ============================================================
# 39. LOGISTIC REGRESSION COEFFICIENTS
# ============================================================

coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_[0]
})


coefficients["Absolute_Coefficient"] = (
    coefficients["Coefficient"].abs()
)


coefficients = coefficients.sort_values(
    by="Absolute_Coefficient",
    ascending=False
)

print("\n================ TOP FEATURES ================")
print(
    coefficients[
        ["Feature", "Coefficient"]
    ].head(15)
)
```

```python
# ============================================================
# 40. TOP FEATURES VISUALIZATION
# ============================================================

top_features = coefficients.head(10)


plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_features,
    x="Coefficient",
    y="Feature"
)

plt.title("Top 10 Features Influencing Customer Churn")
plt.xlabel("Logistic Regression Coefficient")
plt.ylabel("Feature")
plt.show()
```

```python
# ============================================================
# 41. PREDICT ONE CUSTOMER
# ============================================================

# Take one customer from test dataset
sample_customer = X_test.iloc[[0]]
# Scale customer
sample_customer_scaled = scaler.transform(
    sample_customer
)

# Prediction
prediction = model.predict(
    sample_customer_scaled
)

# Probability
probability = model.predict_proba(
    sample_customer_scaled
)[0][1]

print("\n============================================================")
print("              SINGLE CUSTOMER PREDICTION")
print("============================================================")

if prediction[0] == 1:

    print("Prediction: CUSTOMER WILL CHURN")

else:

    print("Prediction: CUSTOMER WILL NOT CHURN")


print(
    f"Churn Probability: {probability * 100:.2f}%"
)
```
